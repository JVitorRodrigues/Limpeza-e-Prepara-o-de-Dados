```python
# IMPORTAÇÕES
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

from google.colab import drive
drive.mount('/content/drive')
```

```python
# NOMES DAS COLUNAS (consulte a documentação do dataset)
colunas = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]

# CARREGAMENTO DOS DADOS
arquivo = '/content/drive/MyDrive/adult/adult.data'

df = pd.read_csv(
    arquivo,
    header=None,
    names=colunas,
    skipinitialspace=True
)

# CONFERÊNCIA DO CARREGAMENTO
print("Dimensões do dataset:", df.shape)

display(df.head())

print("\nTipos de dados:")
print(df.dtypes)
```

```python

# INSPEÇÃO DAS COLUNAS CATEGÓRICAS

colunas_categoricas = df.select_dtypes(include=['object']).columns

for coluna in colunas_categoricas:
    print(f"\nColuna: {coluna}")
    print("Quantidade de categorias:", df[coluna].nunique())
    print("Valores encontrados:")
    print(df[coluna].value_counts(dropna=False).to_string())
```

```python

# PADRONIZAÇÃO DOS TEXTOS (ANTES x DEPOIS)

df_limpo = df.copy()

resultado_padronizacao = []

for coluna in colunas_categoricas:
    categorias_antes = df_limpo[coluna].nunique(dropna=False)

    # Remove espaços extras e padroniza letras minúsculas
    df_limpo[coluna] = (
        df_limpo[coluna]
        .apply(lambda valor: valor.strip().lower() if isinstance(valor, str) else valor)
    )

    categorias_depois = df_limpo[coluna].nunique(dropna=False)

    resultado_padronizacao.append({
        'Coluna': coluna,
        'Categorias antes': categorias_antes,
        'Categorias depois': categorias_depois
    })

comparacao_padronizacao = pd.DataFrame(resultado_padronizacao)

display(comparacao_padronizacao)
```

```python

# CONVERSÃO DOS MARCADORES DE AUSÊNCIA

df_limpo.replace('?', np.nan, inplace=True)

print("Quantidade de valores ausentes por coluna:")
print(df_limpo.isna().sum())

print("\nTotal de valores ausentes no dataset:")
print(df_limpo.isna().sum().sum())
```

```python

# ESCALONAMENTO — MIN-MAX

colunas_numericas = df_limpo.select_dtypes(include=[np.number]).columns

scaler_minmax = MinMaxScaler()

df_minmax = df_limpo.copy()

df_minmax[colunas_numericas] = scaler_minmax.fit_transform(
    df_limpo[colunas_numericas]
)

print("Primeiras linhas após o escalonamento Min-Max:")
display(df_minmax.head())

print("\nValores mínimos:")
print(df_minmax[colunas_numericas].min())

print("\nValores máximos:")
print(df_minmax[colunas_numericas].max())
```

```python

# ESCALONAMENTO — Z-SCORE

scaler_standard = StandardScaler()

df_zscore = df_limpo.copy()

df_zscore[colunas_numericas] = scaler_standard.fit_transform(
    df_limpo[colunas_numericas]
)

print("Primeiras linhas após o escalonamento Z-score:")
display(df_zscore.head())

print("\nMédias das colunas numéricas:")
print(df_zscore[colunas_numericas].mean().round(2))

print("\nDesvios padrão das colunas numéricas:")
print(df_zscore[colunas_numericas].std().round(2))
```

```python

# ESCALONAMENTO — ROBUSTO

scaler_robust = RobustScaler()

df_robusto = df_limpo.copy()

df_robusto[colunas_numericas] = scaler_robust.fit_transform(
    df_limpo[colunas_numericas]
)

print("Primeiras linhas após o escalonamento robusto:")
display(df_robusto.head())

print("\nMedianas das colunas numéricas:")
print(df_robusto[colunas_numericas].median().round(2))

print("\nPrimeiro e terceiro quartis:")
display(df_robusto[colunas_numericas].quantile([0.25, 0.75]).round(2))
```

```python

# COMPARAÇÃO ENTRE AS TÉCNICAS

resultados = []

for coluna in colunas_numericas:
    for tecnica, dados in [
        ('Min-Max', df_minmax),
        ('Z-score', df_zscore),
        ('Robusto', df_robusto)
    ]:
        resultados.append({
            'Coluna': coluna,
            'Técnica': tecnica,
            'Mínimo': dados[coluna].min(),
            'Máximo': dados[coluna].max(),
            'Média': dados[coluna].mean(),
            'Desvio padrão': dados[coluna].std()
        })

comparacao = pd.DataFrame(resultados)

display(comparacao.round(3))
```

```python
# 1. Por que df.isna().sum() pode retornar zero em todas as colunas
# mesmo quando existem dados faltando?

# Porque alguns datasets representam valores desconhecidos com símbolos,
# como '?'. O pandas não reconhece esses símbolos automaticamente como
# valores ausentes. É necessário convertê-los para NaN.

# 2. Na coluna capital-gain, qual técnica é mais afetada pela grande
# quantidade de zeros?

# O Min-Max pode ser bastante afetado pelos valores extremos, pois utiliza
# o mínimo e o máximo da coluna. Como a maioria dos valores de capital-gain
# é zero e existem valores muito altos, muitos registros ficam próximos
# de zero após o escalonamento. O Z-score também pode ser influenciado
# pela distribuição assimétrica. O RobustScaler utiliza mediana e intervalo
# interquartil, mas não elimina a concentração de zeros.

# 3. Faz sentido escalonar todas as colunas numéricas? O que representa fnlwgt?

# Não necessariamente. A coluna fnlwgt representa o peso final de amostragem
# da pessoa, usado para relacionar a observação à população do censo.
# Se ela for utilizada como variável preditora, o escalonamento dependerá
# do modelo e do objetivo da análise. Se for usada como peso estatístico,
# não deve ser tratada automaticamente como uma variável numérica comum.
```
