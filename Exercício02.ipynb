```python
# CARREGAMENTO DOS DADOS

arquivo = '/content/drive/MyDrive/KPMG_VI_New_raw_data_update_final.xlsx'

# Verificar as planilhas disponíveis
excel = pd.ExcelFile(arquivo)
print(excel.sheet_names)

# Carregar a planilha Transactions
df = pd.read_excel(
    arquivo,
    sheet_name='Transactions'
)

display(df.head())
print("Dimensões:", df.shape)
```

```python
# CONFERÊNCIA DO CABEÇALHO E DOS TIPOS

print("Nomes das colunas:")
print(df.columns.tolist())

print("\nTipos de dados:")
print(df.dtypes)

print("\nPrimeiras linhas:")
display(df.head())

print("\nQuantidade de linhas e colunas:")
print(df.shape)
```

```python
# DUPLICATAS EXATAS

duplicatas_exatas = df.duplicated()

print("Quantidade de duplicatas exatas:",
      duplicatas_exatas.sum())

# Visualizar as duplicatas encontradas
display(df[duplicatas_exatas])
```

```python
# DUPLICATAS POR SUBCONJUNTO DE COLUNAS

if 'transaction_id' in df.columns:

    duplicatas_id = df.duplicated(
        subset=['transaction_id'],
        keep=False
    )

    print("Registros com transaction_id repetido:",
          duplicatas_id.sum())

    display(
        df.loc[duplicatas_id].sort_values('transaction_id')
    )

else:
    print("A coluna transaction_id não foi encontrada.")

# Transaction_id deve identificar uma transação individual. Se houver repetições, precisamos investigar se são erros de registro ou alguma particularidade da base antes de remover linhas.
```

```python
# UNICIDADE DOS IDENTIFICADORES

for coluna in ['transaction_id', 'customer_id', 'product_id']:
    if coluna in df.columns:
        print(f"\nColuna: {coluna}")
        print("Valores únicos:", df[coluna].nunique())
        print("Valores ausentes:", df[coluna].isna().sum())
        print("Valores repetidos:",
              df[coluna].duplicated().sum())
```

```python
# ESTRATÉGIA DE TRATAMENTO E IMPACTO

df_limpo = df.copy()

# Indicador antes da limpeza
if 'list_price' in df.columns:
    faturamento_antes = df['list_price'].sum()

# Remover somente linhas completamente duplicadas
quantidade_antes = len(df_limpo)

df_limpo = df_limpo.drop_duplicates()

quantidade_deletada = quantidade_antes - len(df_limpo)

print("Linhas antes:", quantidade_antes)
print("Linhas depois:", len(df_limpo))
print("Duplicatas removidas:", quantidade_deletada)

# Comparar o indicador
if 'list_price' in df.columns:
    faturamento_depois = df_limpo['list_price'].sum()

    print("\nSoma de list_price antes:", faturamento_antes)
    print("Soma de list_price depois:", faturamento_depois)
    print("Diferença:", faturamento_antes - faturamento_depois)
```
